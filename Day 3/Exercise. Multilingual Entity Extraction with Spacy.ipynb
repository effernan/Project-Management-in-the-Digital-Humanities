```python
#Solution

from urllib import request
from bs4 import BeautifulSoup
import re
import pandas as pd
```

We import the data

```python
url = "https://www.gutenberg.org/cache/epub/14155/pg14155.txt"
```

```python
response = request.urlopen(url)
raw = response.read().decode("utf-8")
```

```python
raw
```

```python
print(raw)
```

```python
soup = BeautifulSoup(raw, "html.parser")
data = soup.prettify()
```

```python
print(data)
```

We select the data

```python
text = re.search("START OF THE PROJECT GUTENBERG EBOOK MADAME BOVARY", data)
```

```python
text
```

```python
text_2 = re.search("END OF THE PROJECT GUTENBERG EBOOK", data)
text_2
```

```python
data = data[824:700950]
```

```python
data
```

Let's do some extra cleaning

```python
text = re.search("À Louis Bouilhet", data)
```

```python
text
```

```python
data = data[1008:]
```

```python
data
```

We do some cleaning of the data

```python
data_1 = re.sub("\r\n", "", data)
```

# 2. Spacy

And now let's do NER with Spacy!

```python
#Your code in here
```

```python
#solution
import spacy
```

```python
!python -m spacy download fr_core_news_sm 
```

```python
nlp = spacy.load("fr_core_news_sm")
```

```python
doc = nlp(data_1)
```

```python
nlp.get_pipe('ner').labels
```

We are going to focus on LOC

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
LOC = []

for ent in doc.ents:
    if ent.label_ == "LOC":
        LOC.append([ent.text, ent.label_])
```
